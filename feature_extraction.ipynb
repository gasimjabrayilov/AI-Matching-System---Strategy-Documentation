```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
job_df = pd.read_csv("job_descriptions.csv")
nurse_df = pd.read_csv("nurse_profiles.csv")
```

```python
job_df.head()
```

```python
job_skills = job_df['Job_Description'].fillna("")
nurse_skills = nurse_df['Certifications'].fillna("")
```

```python
combined_text = pd.concat([job_skills, nurse_skills])
```
